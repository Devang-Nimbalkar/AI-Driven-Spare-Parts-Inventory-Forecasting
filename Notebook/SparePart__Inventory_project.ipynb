```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df.info()
```

```python
df["Year"]=df["Date"].dt.year
df["Month"]=df["Date"].dt.month
df["Day"]=df["Date"].dt.day
df["Weekday"]=df["Date"].dt.weekday
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
categorical_cols=["SKU_ID","Warehouse_ID","Supplier_ID","Region"]
for col in categorical_cols:
  le=LabelEncoder()
  df[col]=le.fit_transform(df[col])
```

```python
y=df["Units_Sold"]
```

```python
x=df.drop(["Units_Sold","Date"],axis=1)
```

```python
from sklearn.model_selection import  train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor
model=RandomForestRegressor(
n_estimators=100,
random_state=42)
model.fit(x_train,y_train)
```

```python
predictions=model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
mae=mean_absolute_error(y_test,predictions)
mse=mean_squared_error(y_test,predictions)
r2=r2_score(y_test,predictions)
```

```python
print("MAE:",mae)
print("MSE:",mse)
print("R2:",r2)
```

```python
feature_importance=pd.DataFrame({"Feature":x.columns,"Importance":model.feature_importances_})
feature_importance=feature_importance.sort_values(by="Importance",ascending=False)
```

```python
import matplotlib.pyplot as plt
feature_importance.head(10).plot(x="Feature",y="Importance",kind="bar",figsize=(10,5))
plt.title("Feature Importance")
plt.show()
```

## Inventory Recommendation Engine

```python
def inventory_recommendation(row):
  if row["Inventory_Level"]<= row["Reorder_Point"]:
    return "Reorder Required"
  else:
    return "Inventory Sufficient"
```

```python
df["Recommendation"]=df.apply(inventory_recommendation,axis=1)
```

```python
df[["Inventory_Level","Reorder_Point","Recommendation"]].head(20)
```

```python
df["Recommendation"].value_counts()
```

```python
# Define the list of features that the model was originally trained on.
# These are all columns from `df` after initial preprocessing (date features, label encoding),
# but excluding the target ('Units_Sold'), the original 'Date' column,
# and any columns added *after* `x_train` was created (like 'Recommendation').
trained_features = [
    'SKU_ID', 'Warehouse_ID', 'Supplier_ID', 'Region', 'Inventory_Level',
    'Supplier_Lead_Time_Days', 'Reorder_Point', 'Order_Quantity',
    'Unit_Cost', 'Unit_Price', 'Promotion_Flag', 'Stockout_Flag',
    'Demand_Forecast', 'Year', 'Month', 'Day', 'Weekday'
]

# Create a DataFrame for prediction by selecting only the trained features from the current `df`.
X_for_prediction = df[trained_features]

df["Predicted_Demand"] = model.predict(X_for_prediction)
```

```python
df[["Predicted_Demand"]].head()
```

```python
def smart_recommendation(row):
  if row["Inventory_Level"]< row["Predicted_Demand"]:
    return "Urgent Reorder"
  elif row["Inventory_Level"]<= row["Reorder_Point"]:
    return "Reorder soon"
  else:
    return "Inventory Healthy"
```

```python
df["Smart_Recommendation"]=df.apply(smart_recommendation,axis=1)
```
